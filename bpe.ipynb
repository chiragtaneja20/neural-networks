```python
counts = {} 
def get_stats(ids):
    counts = {} 
    for pair in zip(ids,ids[1:]):
        counts[pair] = counts.get(pair,0) +  1
    return counts 
    
stats = get_stats(x)
print(x) 
print('-------') 
print(stats)
print('-------') 
t = [] 
for k,v in stats.items():
    t.append((v,k)) 
print(sorted(tuple(t),reverse = True))
    
    
```

```python
top_pair = max(stats, key=stats.get) 
top_pair
```

```python
def merge(ids, pair, idx):
    newids = []
    i = 0
    while i < len(ids):
        if i < len(ids) - 1 and ids[i] == pair[0] and ids[i+1] == pair[1]:
            newids.append(idx) 
            i+=2 
        else: 
            newids.append(ids[i])
            i+=1
    return newids 

# here we are trying to replace the top most pair with the new index that will be 256 because in unicode there are only 255 

x2 = merge(x, top_pair, 256)
print(x2)
```

```python
vocab_size = 300
num_merges = vocab_size - 256
ids = list(x) 

merges = {}
for i in range(num_merges):
    stats = get_stats(ids) 
    pair = max(stats, key = stats.get)
    idx = 256 + i
    print(f"merging {pair} into a new token {idx}")
    ids = merge(ids, pair, idx) 
    merges[pair] = idx
```
